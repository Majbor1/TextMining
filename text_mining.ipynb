# Analiza streszczeń literatury młodzieżowej z końca XX wieku.

## Import bibliotek

```python
import json
import nltk
import string
import morfeusz2
import operator as op
import itertools as it
import pandas as pd
from nltk.tokenize import word_tokenize
from nltk.corpus import PlaintextCorpusReader
```

```python
nltk.pathsec.ENFORCE = False
nltk.download('punkt_tab')
```

## Utworzenie korpusu dokumentów

```python
corpus_dir = "./Literature"
corpus = PlaintextCorpusReader(corpus_dir, r".*\.txt")
files_names = corpus.fileids()
```

## Wstępne przygotowanie dokumentów

```python
documets_dict = {}
for file_name in files_names:
    documets_dict[file_name] = corpus.raw(file_name)
print(json.dumps(documets_dict, indent=4, ensure_ascii=False)) 
```

```python
lengths = {}
for file_name in documets_dict:
    lengths[file_name] = {
        "pre": len(word_tokenize(documets_dict[file_name])),
    }
lengths
```

```python
stopwords_file = open("./stopwords_pl.txt")
stopwords = stopwords_file.read().splitlines()
stopwords = stopwords[4:]
stopwords_file.close()
stopwords
```

```python
def lemmatize(text):
    morfer = morfeusz2.Morfeusz()
    segments = it.groupby(morfer.analyse(text), op.itemgetter(0,1))
    def disambiguate(group):
        pairs = ((len(descr), lemma) for _, _, (_, lemma, descr, _, _, ) in group)
        perpl, lemma = min(pairs)
        return lemma.split(":")[0] if lemma else None
    lemmas = (disambiguate(group) for _, group in segments)
    return " ".join(filter(str.isalpha, lemmas))
        
```

```python
for file_name in documets_dict:
    documets_dict[file_name] = documets_dict[file_name].lower()
    documets_dict[file_name] = "".join([char for char in documets_dict[file_name] if char not in string.punctuation])
    documets_dict[file_name] = lemmatize(documets_dict[file_name])
    documets_dict[file_name] = " ".join([word for word in word_tokenize(documets_dict[file_name], language='polish') if word not in stopwords])
print(json.dumps(documets_dict, indent=4, ensure_ascii=False))
```

```python
for file_name in documets_dict:
    lengths[file_name]["post"] = len(word_tokenize(documets_dict[file_name]))
print(json.dumps(lengths, indent=4, ensure_ascii=False))
```

```python
lengths = pd.DataFrame.from_dict(lengths, orient='index')
lengths['change'] = lengths['post'] - lengths['pre']
lengths['change_percent'] = ((lengths['change'] / lengths['pre']) * 100).round(2)
lengths
```
